## 1. Lists — the Basics

A **list** stores many values together, **in order**, inside square brackets
`[ ]`. Lists can hold numbers, text, or a mix — and they can be **changed**.

Make a list and print it:

```python
fruits = ["apple", "banana", "cherry"]
print(fruits)
```

A list can mix different types of values:

```python
mixed = ["Rahul", 21, True, 3.5]
print(mixed)
```

Get one item by its position (**index**), starting at 0:

```python
fruits = ["apple", "banana", "cherry"]
print(fruits[0])    # first
print(fruits[-1])   # last
```

**`len()`** tells you how many items are in the list:

```python
fruits = ["apple", "banana", "cherry"]
print(len(fruits))
```

Join two lists with **`+`** (concatenation):

```python
a = [1, 2, 3]
b = [4, 5]
print(a + b)
```

Repeat a list with **`*`**:

```python
print([0] * 5)
```

Check if a value is in the list with **`in`**:

```python
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)
print("mango" in fruits)
```

Go through every item with a **`for`** loop:

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

## 2. List Slicing

**Slicing** grabs a **piece** of a list using `[start:stop]`. Just like with
strings, the `stop` position is **not included**.

`[1:4]` takes items at positions 1, 2 and 3:

```python
nums = [10, 20, 30, 40, 50]
print(nums[1:4])
```

Leave out the **start** (begin at 0) or the **stop** (go to the end):

```python
nums = [10, 20, 30, 40, 50]
print(nums[:3])    # first three
print(nums[2:])    # from position 2 to the end
```

Use **negative** positions — `[-2:]` gets the last two items:

```python
nums = [10, 20, 30, 40, 50]
print(nums[-2:])
```

Add a **step** with `[start:stop:step]`. A step of 2 takes every second item:

```python
nums = [10, 20, 30, 40, 50]
print(nums[::2])
```

Reverse a list with `[::-1]`:

```python
nums = [10, 20, 30, 40, 50]
print(nums[::-1])
```

## 3. List Methods

Lists come with built-in **methods** (functions you call with a dot) to add,
remove, and organise items. Here are the most common ones.

**`append()`** adds one item to the **end**:

```python
fruits = ["apple", "banana"]
fruits.append("cherry")
print(fruits)
```

**`insert(position, item)`** adds an item at a chosen position:

```python
nums = [1, 2, 4]
nums.insert(2, 3)     # put 3 at position 2
print(nums)
```

**`extend()`** adds **each item** of another list.
(Compare with `append`, which would add the whole list as one item.)

```python
a = [1, 2]
a.extend([3, 4])
print(a)
```

See the difference — **`append`** puts the list inside as a single item:

```python
a = [1, 2]
a.append([3, 4])
print(a)
```

**`remove(value)`** deletes the first matching value:

```python
fruits = ["apple", "banana", "cherry"]
fruits.remove("banana")
print(fruits)
```

**`pop()`** removes the **last** item and **gives it back** to you:

```python
fruits = ["apple", "banana", "cherry"]
removed = fruits.pop()
print("removed:", removed)
print("left:", fruits)
```

**`pop(position)`** removes the item at a chosen position:

```python
fruits = ["apple", "banana", "cherry"]
fruits.pop(0)     # remove the first
print(fruits)
```

**`index(value)`** tells you the position of a value:

```python
letters = ["a", "b", "c"]
print(letters.index("b"))
```

**`count(value)`** counts how many times a value appears:

```python
nums = [1, 2, 2, 3, 2]
print(nums.count(2))
```

**`sort()`** puts the list in order (this changes the list itself):

```python
nums = [3, 1, 4, 1, 5, 2]
nums.sort()
print(nums)
```

**`reverse()`** flips the order of the list:

```python
nums = [1, 2, 3, 4]
nums.reverse()
print(nums)
```

**`clear()`** empties the list completely:

```python
nums = [1, 2, 3]
nums.clear()
print(nums)
```

Handy built-in functions for number lists: **`min`**, **`max`**, **`sum`**:

```python
nums = [4, 1, 7, 3]
print(min(nums))
print(max(nums))
print(sum(nums))
```

### List methods — quick reference

| Method | What it does |
|--------|--------------|
| `append(x)` | add `x` to the end |
| `insert(i, x)` | add `x` at position `i` |
| `extend(list)` | add each item of another list |
| `remove(x)` | delete first `x` |
| `pop()` / `pop(i)` | remove & return last / item `i` |
| `index(x)` | position of `x` |
| `count(x)` | how many `x` |
| `sort()` | put in order |
| `reverse()` | flip the order |
| `clear()` | empty the list |

## 4. Cloning Lists & Mutability

Lists are **mutable** — they can be changed after they are made. That is powerful,
but it leads to one important surprise you must understand.

**Mutable** means you can change an item in place:

```python
fruits = ["apple", "banana", "cherry"]
fruits[0] = "mango"
print(fruits)
```

### ⚠️ The surprise: `b = a` does **not** make a copy

Writing `b = a` gives the **same** list two names. Change one, and the other
changes too — because they are the very same list in memory.

Watch — changing `b` also changes `a`:

```python
a = [1, 2, 3]
b = a            # NOT a copy - same list, two names
b.append(4)
print("a =", a)
print("b =", b)
```

We can prove they are the same list with **`is`**:

```python
a = [1, 2, 3]
b = a
print(a is b)    # True -> the same list
```

### The fix: **clone** (make a real copy)

To get an **independent** copy, use one of these three ways:
`a.copy()`, `a[:]`, or `list(a)`.

**`copy()`** makes a separate list. Now changing `b` leaves `a` alone:

```python
a = [1, 2, 3]
b = a.copy()
b.append(4)
print("a =", a)     # unchanged
print("b =", b)
```

The **slice `[:]`** trick makes a copy too:

```python
a = [1, 2, 3]
b = a[:]
b.append(99)
print("a =", a)
print("b =", b)
```

And **`list()`** makes a copy as well. Here `a is c` is `False` — a real copy:

```python
a = [1, 2, 3]
c = list(a)
print(a is c)    # False -> a different list
```

## 5. Tuples

A **tuple** is like a list, but it uses **round brackets `( )`** and it is
**immutable** — once made, it **cannot be changed**. Use a tuple for values that
should stay fixed, like coordinates or a date.

Make a tuple and read its items with indexing:

```python
point = (3, 4)
print(point[0])
print(point[1])
```

Slicing works on tuples, just like lists:

```python
colors = ("red", "green", "blue", "yellow")
print(colors[1:3])
```

But you **cannot change** a tuple. Trying to raises a `TypeError`
(we catch it so we can read the message):

```python
point = (3, 4)
try:
    point[0] = 5
except TypeError as e:
    print("TypeError:", e)
```

**Careful:** a one-item tuple needs a **comma**, or it is not a tuple at all:

```python
one = (5,)      # a tuple
not_tuple = (5)  # just the number 5
print(type(one))
print(type(not_tuple))
```

**Unpacking** — pull a tuple's values into separate variables in one line:

```python
point = (3, 4)
x, y = point
print("x =", x)
print("y =", y)
```

Tuples have just **two** methods: `count()` and `index()`:

```python
nums = (1, 2, 2, 3)
print(nums.count(2))
print(nums.index(3))
```

### List vs Tuple — quick reminder

| | List | Tuple |
|--|------|-------|
| Brackets | `[ ]` | `( )` |
| Changeable? | ✅ Yes (mutable) | ❌ No (immutable) |
| Best for | data that changes | data that stays fixed |

## 6. Dictionaries

A **dictionary** stores information as **key → value** pairs, inside curly braces
`{ }`. Instead of a position number, you look things up by their **key**
(like looking up a word in a real dictionary).

Make a dictionary and look up a value by its key:

```python
student = {"name": "Rahul", "age": 21}
print(student["name"])
print(student["age"])
```

**Add** a new pair, or **change** an existing value, using the key:

```python
student = {"name": "Rahul", "age": 21}
student["grade"] = "A"     # add a new pair
student["age"] = 22        # change an existing value
print(student)
```

**`get()`** looks up a key safely — if the key is missing it returns `None`
(instead of an error), and you can give your own default:

```python
student = {"name": "Rahul", "age": 21}
print(student.get("name"))
print(student.get("phone"))          # missing -> None
print(student.get("phone", "N/A"))   # missing -> your default
```

**`keys()`**, **`values()`** and **`items()`** show the parts of the dictionary:

```python
student = {"name": "Rahul", "age": 21, "grade": "A"}
print(student.keys())
print(student.values())
print(student.items())
```

**Loop** through a dictionary. Looping gives you the **keys**:

```python
student = {"name": "Rahul", "age": 21}
for key in student:
    print(key, "->", student[key])
```

Use **`.items()`** to get the key **and** value together in the loop:

```python
student = {"name": "Rahul", "age": 21}
for key, value in student.items():
    print(key, "=", value)
```

**`in`** checks whether a **key** exists:

```python
student = {"name": "Rahul", "age": 21}
print("name" in student)
print("phone" in student)
```

Remove a pair with **`del`** or with **`pop(key)`** (which returns the value):

```python
student = {"name": "Rahul", "age": 21, "grade": "A"}
del student["age"]
removed = student.pop("grade")
print("removed:", removed)
print(student)
```
