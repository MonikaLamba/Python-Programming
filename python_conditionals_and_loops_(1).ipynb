## 1. Conditional Statements

A **condition** is a question that is either **True** or **False** (for example,
`age >= 18`). Conditional statements let your program **decide** what to do based
on the answer.

> Two important rules:
> - The line ends with a colon `:`
> - The code to run is **indented** (moved to the right, usually 4 spaces)

**`if`** runs the indented code **only when** the condition is True.
Here `age` is 20, so the message prints:

```python
age = 20
if age >= 18:
    print("You are an adult.")
```

**`if / else`** gives a second path for when the condition is False.
Now `age` is 15, so the `else` block runs instead:

```python
age = 15
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

**`if / elif / else`** checks several conditions in order. Python runs the
**first** one that is True, then stops checking. (`elif` is short for "else if".)

```python
marks = 82

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 40:
    print("Grade C")
else:
    print("Fail")
```

You can combine conditions with **`and`** (both must be True) and
**`or`** (at least one must be True):

```python
age = 20
has_id = True

if age >= 18 and has_id:
    print("Allowed to enter.")
```

Conditions can also be **nested** — an `if` inside another `if`:

```python
logged_in = True
is_admin = True

if logged_in:
    if is_admin:
        print("Welcome, admin!")
    else:
        print("Welcome, user!")
```

## 2. The `for` Loop

A **loop** repeats code. A **`for` loop** repeats **once for each item** in a
collection (like a list, a range of numbers, or the letters of a word).

Loop over a **list** — the variable `fruit` becomes each item in turn:

```python
for fruit in ["apple", "banana", "cherry"]:
    print(fruit)
```

**`range(5)`** gives the numbers `0, 1, 2, 3, 4` (it starts at 0 and stops
**before** 5):

```python
for i in range(5):
    print(i)
```

Give `range` a **start and stop** to choose where it begins and ends.
`range(1, 6)` gives `1, 2, 3, 4, 5`:

```python
for i in range(1, 6):
    print(i)
```

A `for` loop can also go through the **letters of a word**:

```python
for letter in "Hi":
    print(letter)
```

A common pattern — add up numbers using a running total.
`total += n` means `total = total + n`:

```python
total = 0
for n in [10, 20, 30]:
    total += n
print("Total:", total)
```

## 3. The `while` Loop

A **`while` loop** repeats **as long as** a condition stays True. Use it when you
don't know in advance how many times to repeat.

Count from 1 to 5. Each turn prints `count`, then adds 1 to it:

```python
count = 1
while count <= 5:
    print(count)
    count += 1
```

> ⚠️ **Important:** something inside the loop must eventually make the
> condition False (here, `count += 1`). If it never does, the loop runs forever —
> this is called an **infinite loop**. Always remember to update the variable!

A `while` loop is great for "keep going until done". Here we cut a number in
half until it drops below 1:

```python
number = 20
while number >= 1:
    print(number)
    number = number / 2
```

## 4. `break` — Stop a Loop Early

**`break`** immediately **stops** the whole loop, even if there are items left.

Count up, but **stop** as soon as we reach 5. Notice 5, 6, 7... never print:

```python
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

A real use — find the **first** even number in a list, then stop looking:

```python
numbers = [7, 3, 9, 4, 5, 8]

for n in numbers:
    if n % 2 == 0:
        print("First even number:", n)
        break
```

## 5. `continue` — Skip One Turn

**`continue`** skips the **rest of the current turn** and jumps to the next one.
The loop keeps going — it does **not** stop.

Print 1 to 5 but **skip 3**. When `i` is 3, `continue` jumps to the next turn
before it can print:

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

Print only the **even** numbers by skipping the odd ones:

```python
for n in range(1, 11):
    if n % 2 != 0:      # if n is odd...
        continue        # ...skip it
    print(n)
```

### `break` vs `continue` — quick reminder

| Keyword | What it does |
|---------|--------------|
| `break` | **Stops** the whole loop right away |
| `continue` | **Skips** the current turn, then keeps looping |
