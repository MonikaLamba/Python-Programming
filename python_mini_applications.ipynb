## App 1 — Simple Calculator

Takes two numbers and an operator, then prints the answer.

**Concepts used:** variables · data types · operators & operands · `if / elif / else` · comparison

Change `num1`, `num2`, or `operation` and run again to try different sums:

```python
num1 = 12
num2 = 5
operation = "+"     # try "+", "-", "*", "/"
# To type your own: operation = input("Enter + - * or / : ")

if operation == "+":
    result = num1 + num2
elif operation == "-":
    result = num1 - num2
elif operation == "*":
    result = num1 * num2
elif operation == "/":
    result = num1 / num2
else:
    result = "Unknown operation"

print(num1, operation, num2, "=", result)
```

## App 2 — Student Average & Grade

Adds up a student's marks, finds the average, and decides the grade.

**Concepts used:** list · `for` loop · assignment (`+=`) · division · `if / elif / else`

Try changing the marks and see how the grade changes:

```python
marks = [85, 78, 92]     # three subject marks

total = 0
for m in marks:
    total += m           # add each mark to the running total

average = total / len(marks)
print("Total  :", total)
print("Average:", average)

if average >= 90:
    print("Grade  : A")
elif average >= 75:
    print("Grade  : B")
elif average >= 40:
    print("Grade  : C")
else:
    print("Grade  : Fail")
```

## App 3 — Number Checker

Tells you whether a number is even or odd, **and** whether it is positive,
negative, or zero.

**Concepts used:** modulus `%` operator · `if / else` · nested conditions · `if / elif / else`

Change `number` to test different values (try a negative one, or 0):

```python
number = 7
# To type your own: number = int(input("Enter a number: "))

# Even or odd (using the remainder operator %)
if number % 2 == 0:
    print(number, "is Even")
else:
    print(number, "is Odd")

# Positive, negative or zero
if number > 0:
    print(number, "is Positive")
elif number < 0:
    print(number, "is Negative")
else:
    print("The number is Zero")
```

## App 4 — Multiplication Table

Prints the full times-table for any number.

**Concepts used:** `for` loop · `range()` · multiplication · precedence (`n * i` before the text) · f-string

Change `n` to print a different table:

```python
n = 7
# To type your own: n = int(input("Table of which number? "))

for i in range(1, 11):
    print(f"{n} x {i} = {n * i}")
```

## App 5 — Add Up 1 to N (while loop)

Adds all the whole numbers from 1 up to N.

**Concepts used:** `while` loop · assignment (`+=`) · comparison · updating the counter

Change `N` to add up more (or fewer) numbers:

```python
N = 10
# To type your own: N = int(input("Add numbers up to: "))

total = 0
i = 1
while i <= N:
    total += i          # add the current number
    i += 1              # move to the next number (very important!)

print(f"The sum of 1 to {N} is {total}")
```

## App 6 — Prime Number Check

A prime number can only be divided by 1 and itself. As soon as we find any other
divider, we can **stop early** with `break`.

**Concepts used:** `for` loop · `break` · modulus `%` · `if / else` · boolean variable

Change `num` to test whether it is prime:

```python
num = 29
# To type your own: num = int(input("Enter a number: "))

is_prime = True

if num < 2:
    is_prime = False
else:
    for i in range(2, num):
        if num % i == 0:      # found a divider
            is_prime = False
            break             # no need to keep checking

if is_prime:
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```

## App 7 — Average of Valid Scores (continue)

Some scores in the list are invalid (negative). We **skip** those with `continue`
and average only the good ones.

**Concepts used:** list · `for` loop · `continue` · comparison · assignment · division

A negative score means "not entered". `continue` skips it:

```python
scores = [50, -1, 80, 90, -5, 70]

total = 0
count = 0
for s in scores:
    if s < 0:               # invalid score
        continue            # skip it and go to the next one
    total += s
    count += 1

print("Valid scores :", count)
print("Average score:", total / count)
```

## App 8 — Voting Eligibility

Checks two things — age and citizenship — to decide if a person can vote.

**Concepts used:** boolean data type · nested `if / else` · logical operators (`and`)

Change `age` and `is_citizen` to test different people:

```python
age = 20
is_citizen = True

if age >= 18:
    if is_citizen:
        print("You are eligible to vote.")
    else:
        print("You must be a citizen to vote.")
else:
    print("You are too young to vote.")

# The same check written with a logical AND:
if age >= 18 and is_citizen:
    print("(Short version: eligible!)")
```

## App 9 — Temperature Converter

Turns a temperature typed as **text** into a number, then converts Celsius to
Fahrenheit.

**Concepts used:** type-conversion function `float()` · operators · precedence (`* / +`) · f-string

Imagine `celsius_text` came from the user as text. We convert it with `float()`,
then apply the formula (Python does `9 / 5`, then `* celsius`, then `+ 32`):

```python
celsius_text = "37"        # text, as if typed by a user
# To type your own: celsius_text = input("Enter temperature in Celsius: ")

celsius = float(celsius_text)          # convert text -> number
fahrenheit = celsius * 9 / 5 + 32      # precedence handles the order

print(f"{celsius} C = {fahrenheit} F")
```

## App 10 — Number Facts (mini capstone)

One small program that uses **many** ideas at once: it looks at a number and
prints several facts about it.

**Concepts used:** variables · conversion · `%` and comparison operators ·
`if / elif / else` · `for` loop · `continue` · f-string

Change `number` and run to get a full mini-report:

```python
number = 12
# To type your own: number = int(input("Enter a number: "))

print(f"Facts about {number}")
print("-" * 20)

# 1) Even or odd
print("Even" if number % 2 == 0 else "Odd")

# 2) Sign
if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")

# 3) All the factors (numbers that divide it exactly)
print("Factors:", end=" ")
for i in range(1, number + 1):
    if number % i != 0:
        continue            # skip non-factors
    print(i, end=" ")
print()
```
