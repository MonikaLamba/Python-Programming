# NumPy — 1-D Arrays (Beginner Friendly)

*The NumPy Library · lots of examples*

**NumPy** (short for *Numerical Python*) is a library for doing fast maths on lots
of numbers at once. Its main tool is the **array**. A **1-D array** is a single row
of numbers — like a Python list, but faster and able to do maths on every element
at the same time.

First, **import** NumPy. Almost everyone gives it the short nickname `np`:

```python
import numpy as np
print("NumPy is ready! Version:", np.__version__)
```

## 1. Creating 1-D Arrays

There are several handy ways to make a 1-D array.

The most common way — turn a **list** into an array with `np.array()`:

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr)
```

It is a special NumPy type called an **ndarray**:

```python
arr = np.array([10, 20, 30, 40, 50])
print(type(arr))
```

**`np.arange(start, stop, step)`** makes a range of numbers
(the `stop` value is not included):

```python
print(np.arange(0, 10, 2))
```

**`np.zeros(n)`** and **`np.ones(n)`** make arrays filled with 0s or 1s:

```python
print(np.zeros(5))
print(np.ones(4))
```

**`np.full(n, value)`** fills an array with any value you choose:

```python
print(np.full(5, 7))
```

**`np.linspace(start, stop, num)`** gives `num` **evenly spaced** numbers
(here the `stop` value *is* included):

```python
print(np.linspace(0, 1, 5))
```

You can make a **random** array too. Setting a *seed* first makes the result
repeatable, so you get the same numbers every run:

```python
np.random.seed(42)
print(np.random.randint(1, 10, size=5))
```

## 2. Array Information

Every array carries useful details about itself.

`ndim` = number of dimensions (1 for a 1-D array), `size` = how many elements,
`dtype` = the type of the numbers inside:

```python
arr = np.array([10, 20, 30, 40, 50])
print("ndim :", arr.ndim)
print("size :", arr.size)
print("dtype:", arr.dtype)
```

**`len()`** also gives the number of elements in a 1-D array:

```python
arr = np.array([10, 20, 30, 40, 50])
print(len(arr))
```

## 3. Accessing Elements

You reach into an array just like a list — with indexing and slicing — plus a few
extra NumPy tricks.

Get **one element** by its position (starts at 0). Negative counts from the end:

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[0])     # first
print(arr[-1])    # last
```

**Slice** a piece with `[start:stop]` (the `stop` position is not included):

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[1:4])
```

**Steps** and **reversing** work too:

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[::2])     # every second element
print(arr[::-1])    # reversed
```

**Fancy indexing** — grab several positions at once using a list of indexes:

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[[0, 2, 4]])
```

Arrays are **changeable** — set a single element to a new value:

```python
arr = np.array([10, 20, 30, 40, 50])
arr[0] = 100
print(arr)
```

You can even change a **whole slice** in one go:

```python
arr = np.array([10, 20, 30, 40, 50])
arr[1:3] = 0
print(arr)
```

## 4. Mathematical Operators (element-wise)

This is NumPy's superpower: an operation applies to **every element at once**, with
**no loop needed**. This is called *element-wise* (or *vectorised*) maths.

Add a number to **every** element at once:

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr + 10)
```

Subtract, multiply, and divide — all element-wise:

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr - 1)
print(arr * 2)
print(arr / 2)
```

Powers `**` and remainders `%` work the same way:

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr ** 2)
print(arr % 2)
```

Add **two arrays** — matching positions are combined:

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
print(a + b)
```

See the difference from a normal list: `*` **repeats** a list, but
**multiplies** an array:

```python
py_list = [1, 2, 3]
arr = np.array([1, 2, 3])
print(py_list * 2)    # list is repeated
print(arr * 2)        # array values are multiplied
```

## 5. Useful Functions

NumPy has fast built-in functions for common calculations.

**Totals and averages** with `np.sum` and `np.mean`:

```python
arr = np.array([4, 8, 15, 16, 23, 42])
print("Sum :", np.sum(arr))
print("Mean:", np.mean(arr))
```

**Biggest, smallest, middle** with `np.max`, `np.min`, `np.median`:

```python
arr = np.array([4, 8, 15, 16, 23, 42])
print("Max   :", np.max(arr))
print("Min   :", np.min(arr))
print("Median:", np.median(arr))
```

**`np.sqrt()`** finds the square root of **every** element:

```python
arr = np.array([1, 4, 9, 16])
print(np.sqrt(arr))
```

**`np.sort()`** returns the array neatly in order:

```python
arr = np.array([3, 1, 4, 1, 5, 9, 2])
print(np.sort(arr))
```

Many functions also work as **methods** (called with a dot):

```python
arr = np.array([4, 8, 15])
print(arr.sum())
print(arr.mean())
print(arr.max())
```

**`np.argmax` / `np.argmin`** give the **position** of the biggest / smallest
value (not the value itself):

```python
arr = np.array([4, 8, 15, 16, 23, 42])
print("Position of max:", np.argmax(arr))
print("Position of min:", np.argmin(arr))
```

## 6. Relational (Comparison) Operators

Comparing an array with a value checks **every element** and returns an array of
`True`/`False`. This is the key to **filtering** data.

**`arr > 10`** checks each element and answers `True` or `False` for each one:

```python
arr = np.array([5, 12, 18, 3, 25])
print(arr > 10)
```

Every comparison works this way — `==`, `<=`, and so on:

```python
arr = np.array([5, 12, 18, 3, 25])
print(arr == 18)
print(arr <= 5)
```

**Boolean indexing** — keep only the elements that pass the test.
Put the condition inside the brackets:

```python
arr = np.array([5, 12, 18, 3, 25])
print(arr[arr > 10])
```

**Combine conditions** with `&` (and) and `|` (or).
Wrap each condition in parentheses `( )`:

```python
arr = np.array([5, 12, 18, 3, 25])
print(arr[(arr > 5) & (arr < 20)])
```

**Count** how many elements pass a test. `True` counts as 1, so `np.sum` works:

```python
arr = np.array([5, 12, 18, 3, 25])
print(np.sum(arr > 10))
```

**`np.any`** asks "is at least one True?" and **`np.all`** asks "are they all
True?":

```python
arr = np.array([5, 12, 18, 3, 25])
print(np.any(arr > 20))
print(np.all(arr > 0))
```
