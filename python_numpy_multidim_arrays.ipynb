# NumPy — Multidimensional Arrays (Beginner Friendly)

*The NumPy Library · 2-D arrays · lots of examples*

A **multidimensional array** has more than one row. The most common is the
**2-D array** — a grid of **rows and columns**, just like a table or a maths
*matrix*:

```
        col0  col1  col2
 row0 [   1     2     3  ]
 row1 [   4     5     6  ]
```

First, **import** NumPy with its usual nickname `np`:

````python
import numpy as np
print("NumPy is ready! Version:", np.__version__)
````

## 1. Creating 2-D Arrays

A 2-D array is made from a **list of lists** — each inner list is one **row**.

Make a 2-D array from nested lists. Notice how it prints as a neat grid:

````python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print(arr)
````

**`np.zeros((rows, cols))`** and **`np.ones((rows, cols))`** — note the shape is
given as a pair `(rows, cols)`:

````python
print(np.zeros((2, 3)))
print(np.ones((3, 3)))
````

**`np.full((rows, cols), value)`** fills a grid with any value:

````python
print(np.full((2, 2), 7))
````

**`np.eye(n)`** makes an *identity matrix* — 1s on the diagonal, 0s elsewhere:

````python
print(np.eye(3))
````

**`reshape`** turns a flat range of numbers into a grid.
Here 12 numbers become 3 rows × 4 columns:

````python
print(np.arange(12).reshape(3, 4))
````

A **random** 2-D array (a seed makes the numbers repeatable):

````python
np.random.seed(0)
print(np.random.randint(1, 10, (2, 3)))
````

## 2. Array Information

For 2-D arrays, `shape` is the most useful detail — it tells you the number of
**rows and columns**.

`ndim` = number of dimensions (2), `shape` = `(rows, cols)`, `size` = total
number of elements:

````python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print("ndim :", arr.ndim)
print("shape:", arr.shape)
print("size :", arr.size)
print("dtype:", arr.dtype)
````

## 3. Accessing Elements

To reach one cell in a 2-D array you give **two positions**: `arr[row, column]`
(both start at 0).

Get single cells with `arr[row, col]`:

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])
print(arr[0, 0])    # row 0, col 0
print(arr[1, 2])    # row 1, col 2
print(arr[-1, -1])  # last row, last col
````

Get a **whole row** by giving just the row number:

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])
print(arr[1])       # the second row
````

Get a **whole column** with `arr[:, col]` — the `:` means "all rows":

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])
print(arr[:, 0])    # the first column
````

**Slice a smaller grid** with `arr[row_start:row_stop, col_start:col_stop]`:

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])
print(arr[0:2, 1:3])   # first two rows, last two columns
````

Arrays are **changeable** — set one cell to a new value:

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60]])
arr[0, 0] = 100
print(arr)
````

You can change a **whole row** (or column) at once:

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60]])
arr[1] = [0, 0, 0]
print(arr)
````

## 4. Mathematical Operators

Just like 1-D arrays, maths happens on **every element at once** — no loops needed.

Add or multiply a number to **every** element (this is called *broadcasting*):

````python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print(arr + 10)
print(arr * 2)
````

Add **two arrays of the same shape** — matching cells are combined:

````python
a = np.array([[1, 2],
              [3, 4]])
b = np.array([[10, 20],
              [30, 40]])
print(a + b)
````

**Broadcasting** a row: a 1-D row is added to **every** row of the grid:

````python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
row = np.array([10, 20, 30])
print(arr + row)
````

**Important:** `*` multiplies **cell by cell**, while `@` does true
**matrix multiplication** — they are different!

````python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])
print("Element-wise (*):")
print(A * B)
print("Matrix multiply (@):")
print(A @ B)
````

**Transpose** with `.T` flips rows into columns:

````python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print(arr.T)
````

## 5. Useful Functions (and the `axis` idea)

Functions like `sum` and `mean` work on the whole grid — but with 2-D arrays you
can also apply them **down columns** or **across rows** using **`axis`**:

```
axis=0   ↓  goes DOWN the rows   → one result per COLUMN
axis=1   →  goes ACROSS columns  → one result per ROW
```

With **no axis**, the function uses the whole grid:

````python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print("Sum of everything:", np.sum(arr))
````

**`axis=0`** works down each **column** (gives one number per column):

````python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print(np.sum(arr, axis=0))
````

**`axis=1`** works across each **row** (gives one number per row):

````python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print(np.sum(arr, axis=1))
````

The same axis idea works for `mean`, `max`, `min`, and more:

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60]])
print("Column means:", np.mean(arr, axis=0))
print("Row maxes   :", np.max(arr, axis=1))
````

**`np.sqrt`** works on every element, and **`flatten()`** turns a 2-D grid back
into a 1-D array:

````python
arr = np.array([[1, 4],
                [9, 16]])
print(np.sqrt(arr))
print("Flattened:", arr.flatten())
````

## 6. Relational (Comparison) Operators

Comparing a 2-D array with a value checks **every cell** and returns a grid of
`True`/`False`. This is the key to **filtering** a table of data.

**`arr > 25`** checks every cell and returns a matching grid of True/False:

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60]])
print(arr > 25)
````

**Boolean indexing** — pull out only the values that pass the test.
The result is a flat 1-D array of the matches:

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60]])
print(arr[arr > 25])
````

**Combine conditions** with `&` (and) and `|` (or), each in parentheses:

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60]])
print(arr[(arr > 15) & (arr < 55)])
````

**Count** how many cells pass a test (`True` counts as 1):

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60]])
print(np.sum(arr > 25))
````

**`np.any`** ("is any True?") and **`np.all`** ("are all True?") — and they take
an `axis` too:

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60]])
print(np.any(arr > 55))          # anywhere in the grid?
print(np.all(arr > 5))           # every cell?
print(np.all(arr > 5, axis=1))   # per row
````
