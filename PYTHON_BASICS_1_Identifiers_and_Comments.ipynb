Comments are notes or explanations written inside a Python program that are not executed by the Python interpreter.

Comments are mainly used to:

Explain what the code does
Make programs easier to understand
Improve code readability
Help in debugging and maintenance
Provide information to other programmers

```python
# This program calculates the sum of two numbers

a = 10
b = 20

print(a + b)
```

Here, the line beginning with # is a comment. Python ignores it during execution.

Single-Line Comments: A single-line comment begins with the # symbol.

```python
# This is a comment
name = "Rahul"
print(name)
```

Everything written after # on that line is treated as a comment.

Inline Comment

A comment can also be written after a statement:

```python
age = 20  # Student's age
print(age)
```

Multi-Line Comments: Python does not have a separate special syntax specifically called a "multi-line comment." However, multiple lines beginning with # can be used.

```python
# This program calculates
# the average marks of
# a student.

marks = 80
print(marks)
```

Triple Quotes and Comments

Triple-quoted strings can span multiple lines:

```python
"""
This is a multi-line string.
It can contain several lines.
"""
Subject = "Python"  # here "" is important for strings
print(Subject)
```

However, technically this is not a comment. It is a string literal that is not assigned to a variable.

Triple-quoted strings are commonly used for docstrings, which document functions, classes, and modules.

```python
def calculate_sum(a, b):
    """This function returns the sum of two numbers."""
    return a + b
```

Here, the text inside """ """ is a docstring.

Why are Comments Important?

1. Improve Readability

Comments help programmers understand the purpose of code.
# Calculate total marks
total = maths + science + english

2. Explain Complex Logic
# Calculate discount only if purchase exceeds ₹5,000
if amount > 5000:
    discount = amount * 0.10

3. Help During Debugging

A programmer can temporarily prevent a line from executing by commenting it:

# print("Testing this statement")

4. Help Team Members

When several programmers work on the same project, comments make the code easier for everyone to understand.

5. Comments vs. Code

Consider:

# Store student's marks
marks = 85

# Display the marks
print(marks)

Python executes:

marks = 85
print(marks)

It does not execute:

# Store student's marks
# Display the marks

6. Good Practices for Writing Comments

Write clear and meaningful comments.
Keep comments short and relevant.
Explain why something is being done when the reason is not obvious.
Avoid comments that simply repeat the code.

❌ Unnecessary Comment
x = 10  # Assign 10 to x

✅ More Useful Comment

# Maximum number of attempts allowed
max_attempts = 10
