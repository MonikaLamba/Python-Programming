## 1. Precedence of Operators

When an expression has more than one operator, Python needs to know which one to
do **first**. That order is called **precedence** (just like *BODMAS* in maths).

**Multiply before add.** In `2 + 3 * 4`, Python does `3 * 4` first, then adds 2:

```python
print(2 + 3 * 4)
```

**Parentheses `( )` change the order** — whatever is inside is done first:

```python
print((2 + 3) * 4)
```

**Power `**` runs before multiply/divide.** Here `4 ** 2 = 16` happens first,
then `3 * 16`:

```python
print(3 * 4 ** 2)
```

**Same-level operators go left to right.** `10 - 4 - 3` means `(10 - 4) - 3`:

```python
print(10 - 4 - 3)
```

**But power `**` goes right to left** (a special case).
`2 ** 3 ** 2` means `2 ** (3 ** 2)` = `2 ** 9`:

```python
print(2 ** 3 ** 2)
```

**Maths is done before comparisons.** First `5 + 3 = 8`, then check `8 > 6`:

```python
print(5 + 3 > 6)
```

**`and` is done before `or`.**
`True or False and False` means `True or (False and False)`:

```python
print(True or False and False)
```

### The precedence order (highest first)

| Level | Operators | Meaning |
|-------|-----------|---------|
| 1 (top) | `( )` | parentheses |
| 2 | `**` | power |
| 3 | `-x  +x  ~x` | unary sign (a minus in front of one number) |
| 4 | `*  /  //  %` | multiply, divide, floor-divide, remainder |
| 5 | `+  -` | add, subtract |
| 6 | `<  <=  >  >=` | comparisons |
| 7 | `==  !=` | equal / not equal |
| 8 | `not` | logical NOT |
| 9 | `and` | logical AND |
| 10 (bottom) | `or` | logical OR |

> **Tip:** If you are ever unsure, add parentheses. They make your intent clear
> *and* guarantee the order you want.

## 2. Type Conversion Functions

**Type conversion** means changing a value from one data type into another — for
example turning the text `"25"` into the number `25`. It is also called **type
casting**, and you do it with built-in functions like `int()`, `float()`, `str()`.

**`int()` makes a whole number.** Text that looks like a number can become a
real number you can calculate with:

```python
print(int("25"))
print(int("25") + 5)
```

**`int()` on a decimal cuts off the fraction** — it does *not* round:

```python
print(int(3.9))
print(int(-3.9))
```

**`float()` makes a decimal number:**

```python
print(float("3.14"))
print(float(7))
```

**`str()` makes text.** Handy when you want to join a number with words:

```python
age = 21
print(str(age))
print("Age is " + str(age))
```

**`bool()` makes a True/False value.** Empty things and zero are `False`;
everything else is `True`:

```python
print(bool(0))
print(bool(1))
print(bool(""))
print(bool("hi"))
```

**Careful!** The text `"False"` is *not empty*, so it becomes `True`:

```python
print(bool("False"))
```

**`list()`, `tuple()`, `set()` change between collections**
(and `set()` also removes duplicates):

```python
print(list("abc"))
print(tuple([1, 2, 3]))
print(set([1, 1, 2, 3, 3]))
```

**Automatic (implicit) conversion.** Sometimes Python converts for you.
Mixing an `int` and a `float` gives a `float`:

```python
print(5 + 2.0)
```

**When conversion is not possible, you get an error.**
`int("3.5")` fails because `"3.5"` is not a whole number as text.
We can catch it safely with `try` / `except`, then fix it by going through `float` first:

```python
try:
    value = int("3.5")
except ValueError as e:
    print("ValueError:", e)
    print("Fix:", int(float("3.5")))
```

## 3. Short-Circuit vs Lazy Evaluation

**Big idea:** Python is **lazy** — it does only as much work as it needs, and no more.

- **Short-circuit** evaluation is this laziness applied to `and` / `or`:
  Python stops the moment it already knows the answer.
- **Lazy** evaluation is the wider idea: values are produced only **when needed**
  (for example, `range()` and generators don't build everything up front).

### Part A — Short-circuit ( `and` / `or` )

First we make a little helper that **prints when it is used**, so we can *see*
exactly what Python checks. Run this cell first:

```python
def check(label, value):
    print(f"   (Python actually looked at {label})")
    return value
```

**`and` stops early if the first value is `False`.**
The whole `and` is already `False`, so the right side is skipped —
notice only the left side prints:

```python
result = check("left", False) and check("right", True)
print("result =", result)
```

**`or` stops early if the first value is `True`.**
The whole `or` is already `True`, so the right side is skipped:

```python
result = check("left", True) or check("right", False)
print("result =", result)
```

**`and` / `or` return the actual value, not just True/False.**
`and` gives the first falsy value (or the last value if all are true);
`or` gives the first truthy value:

```python
print(0 and 5)
print(3 and 5)
print(0 or 5)
print(3 or 5)
```

**Why it's useful: avoid errors safely.** Dividing by zero crashes, but because
`x != 0` is `False`, the division on the right is never even tried — no error:

```python
x = 0
safe = (x != 0) and (10 / x > 1)
print(safe)
```

### Part B — Lazy evaluation (produce values only when needed)

**`range()` is lazy** — it does *not* build a giant list. `range(1, 1000001)`
describes a million numbers but creates each one only when asked, so it stays tiny:

```python
numbers = range(1, 1000001)
print(type(numbers).__name__)
print(numbers[0])
print(numbers[-1])
```

**A generator produces values one at a time.** The `( )` below make a generator;
it computes each square only when we ask for the next one with `next()`:

```python
squares = (n * n for n in range(1, 6))
print(next(squares))
print(next(squares))
print(next(squares))
```

### Putting it together

- **Short-circuit** = laziness for `and` / `or` (stop as soon as the answer is known).
- **Lazy sequences** = laziness for data (make values only when needed).

Both follow the same rule: **do the least work necessary.**
