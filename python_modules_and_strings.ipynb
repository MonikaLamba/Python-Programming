## 1. Importing Modules

A **module** is a ready-made file of useful code — functions and values someone
already wrote for you. Python comes with many built-in modules. To use one, you
**import** it first with the `import` keyword.

Import the **`math`** module, then use its `sqrt()` function to get a square root.
Notice we write `math.` before the function name:

````python
import math
print(math.sqrt(16))
````

The `math` module also stores handy **values**, like `pi`:

````python
print(math.pi)
````

Round a number **down** with `floor()` and **up** with `ceil()`:

````python
print(math.floor(4.3))
print(math.ceil(4.3))
````

A few more `math` tools — power and factorial:

````python
print(math.pow(2, 5))      # 2 to the power 5
print(math.factorial(5))   # 5 x 4 x 3 x 2 x 1
````

You can import **just one function** using `from ... import`.
Then you use it **without** the `math.` prefix:

````python
from math import sqrt
print(sqrt(25))
````

Import **several names** at once, separated by commas:

````python
from math import pi, floor
print(pi)
print(floor(9.8))
````

Give a module a short **nickname** with `as` (this is called an *alias*).
Here `math` becomes `m`:

````python
import math as m
print(m.sqrt(81))
````

> **Note:** You may see `from math import *` which brings in *everything* at
> once. It looks convenient but can cause name clashes, so it's best avoided.
> Prefer `import math` or `from math import sqrt`.

The **`random`** module makes random numbers. `randint(1, 6)` gives a whole
number from 1 to 6 (both included) — like rolling a dice:

````python
import random
print(random.randint(1, 6))
````

*(Your number will be different each time you run the cell — that's the point
of random!)*

`choice()` picks a **random item** from a list:

````python
import random
colors = ["red", "green", "blue"]
print(random.choice(colors))
````

The **`datetime`** module can tell you **today's date**:

````python
import datetime
print(datetime.date.today())
````

The **`statistics`** module does quick maths on a list of numbers:

````python
import statistics
marks = [70, 85, 90, 60]
print(statistics.mean(marks))     # average
print(statistics.median(marks))   # middle value
````

### Three ways to import — quick summary

| Way to write it | How you use it |
|-----------------|----------------|
| `import math` | `math.sqrt(16)` |
| `from math import sqrt` | `sqrt(16)` |
| `import math as m` | `m.sqrt(16)` |

## 2. Strings — Basic Functions

A **string** is text, written inside quotes (`"..."` or `'...'`). Strings come
with many built-in **functions** (called **methods**) that you use with a dot,
like `text.upper()`. Let's explore them with plenty of examples.

Make a string, and find how many characters it has with **`len()`**:

````python
name = "Python"
print(name)
print(len(name))
````

**`upper()`** makes everything CAPITALS, **`lower()`** makes everything small:

````python
text = "Hello World"
print(text.upper())
print(text.lower())
````

**`title()`** capitalises Each Word; **`capitalize()`** capitalises only the
first letter of the string:

````python
text = "hello world"
print(text.title())
print(text.capitalize())
````

**`strip()`** removes extra spaces from both ends (very useful for tidying
user input):

````python
messy = "   hello   "
print(messy.strip())
````

**`replace(old, new)`** swaps one piece of text for another:

````python
sentence = "I like cats"
print(sentence.replace("cats", "dogs"))
````

**`split()`** breaks a sentence into a **list of words**:

````python
sentence = "learn python step by step"
print(sentence.split())
````

**`join()`** does the opposite — it glues a list of words back into one
string. The string before `.join` is the "glue" (here, a space):

````python
words = ["learn", "python", "today"]
print(" ".join(words))
````

**`count()`** tells you how many times something appears:

````python
text = "banana"
print(text.count("a"))
````

**`find()`** gives the position where text first appears
(and returns `-1` if it is not found):

````python
text = "hello"
print(text.find("l"))    # found at position 2
print(text.find("z"))    # not found -> -1
````

**`startswith()`** and **`endswith()`** check the beginning and end —
great for checking file names:

````python
filename = "report.pdf"
print(filename.startswith("report"))
print(filename.endswith(".pdf"))
````

**Content checks** — is it all digits? all letters? letters-or-digits?

````python
print("12345".isdigit())
print("hello".isalpha())
print("hello123".isalnum())
````

Check if some text is **inside** another string using **`in`**:

````python
sentence = "I love Python"
print("Python" in sentence)
print("Java" in sentence)
````

### Indexing and slicing

Every character has a **position number** starting at **0**. Use square brackets
`[ ]` to grab characters.

```
 P  y  t  h  o  n
 0  1  2  3  4  5     <- positions
-6 -5 -4 -3 -2 -1     <- positions from the end
```

Get **one character** by its position (counting starts at 0):

````python
word = "Python"
print(word[0])    # first character
print(word[1])    # second character
````

**Negative** positions count from the end. `-1` is the last character:

````python
word = "Python"
print(word[-1])
````

Grab a **slice** (a piece) with `[start:stop]`. It stops **before** `stop`,
so `[0:3]` gives characters 0, 1 and 2:

````python
word = "Python"
print(word[0:3])
````

A neat trick — **reverse** a string with `[::-1]`:

````python
word = "Python"
print(word[::-1])
````

### Joining and repeating

Join strings together with **`+`** (this is called *concatenation*):

````python
first = "Data"
second = "Structure"
print(first + " " + second)
````

**Repeat** a string with **`*`** — handy for drawing lines:

````python
print("ha" * 3)
print("-" * 20)
````

Put variables neatly inside text using an **f-string** — just add an `f`
before the opening quote and wrap variables in `{ }`:

````python
name = "Aisha"
age = 21
print(f"My name is {name} and I am {age} years old.")
````

> **Good to know:** string methods return a **new** string — they do **not**
> change the original. See below: `text` is still lowercase afterwards.

Proof that the original string is unchanged:

````python
text = "hello"
print(text.upper())   # HELLO  (a new string)
print(text)           # hello  (original is untouched)
````

### Common string methods — quick reference

| Method | What it does |
|--------|--------------|
| `len(s)` | number of characters |
| `s.upper()` / `s.lower()` | change case |
| `s.title()` / `s.capitalize()` | capitalise words / first letter |
| `s.strip()` | remove spaces at the ends |
| `s.replace(a, b)` | swap `a` for `b` |
| `s.split()` | text → list of words |
| `" ".join(list)` | list of words → text |
| `s.count(x)` | how many times `x` appears |
| `s.find(x)` | position of `x` (or `-1`) |
| `s.startswith(x)` / `s.endswith(x)` | check start / end |
| `x in s` | is `x` inside `s`? |
