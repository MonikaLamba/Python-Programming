## App 1 — Username Validator

Checks a username against a few rules before allowing it.

**Concepts used:** strings (`len`, subscript `[0]`, `isalpha`) · `if / elif / else` · `not` · comparison

Rules: 5–15 characters, and it must start with a letter. Change `username` to test:

```python
username = "Rahul_23"
# To type your own: username = input("Choose a username: ")

if len(username) < 5:
    print("Too short (need at least 5 characters).")
elif len(username) > 15:
    print("Too long (max 15 characters).")
elif not username[0].isalpha():
    print("Must start with a letter.")
else:
    print("Valid username!")
```

## App 2 — Text Analyzer

Counts the characters, words, and vowels in a sentence.

**Concepts used:** string methods (`len`, `split`, `lower`) · `for` loop · `in` · `if` · assignment

Change the `sentence` and run to analyse different text:

```python
sentence = "Python is fun and Python is powerful"

print("Characters:", len(sentence))

words = sentence.split()
print("Words     :", len(words))

vowels = "aeiou"
count = 0
for ch in sentence.lower():
    if ch in vowels:
        count += 1
print("Vowels    :", count)
```

## App 3 — Palindrome Checker

A palindrome reads the same forwards and backwards (like *madam*). We reverse the
word with slicing and compare.

**Concepts used:** string slicing `[::-1]` · `lower` · comparison · `if / else` · immutability (we build a **new** reversed string)

Try `"Madam"`, `"level"`, or `"python"`:

```python
word = "Madam"
# To type your own: word = input("Enter a word: ")

clean = word.lower()          # make case not matter
reversed_word = clean[::-1]   # a NEW reversed string (original is untouched)

if clean == reversed_word:
    print(word, "is a palindrome")
else:
    print(word, "is not a palindrome")
```

## App 4 — Name Formatter

Turns a full name into initials and a neat title-case version.

**Concepts used:** `split` · subscript `[0]` · `upper` · `title` · assignment (`+=`) · f-string · building a new string (immutability)

Change `full_name` and run:

```python
full_name = "rahul kumar sharma"

parts = full_name.split()      # a list of the words
initials = ""
for p in parts:
    initials += p[0].upper() + "."   # first letter of each word

print("Initials    :", initials)
print("Proper name :", full_name.title())
```

## App 5 — Shopping Cart & Bill

Manages a cart with list **methods**, then totals the bill by **converting** text
prices into numbers.

**Concepts used:** list methods (`append`, `remove`, `sort`) · `len` · type-conversion `int()` · `for` · `+=`

Add and remove items, then add up the prices:

```python
cart = ["milk", "bread", "eggs"]
cart.append("butter")     # add an item
cart.remove("bread")      # remove an item
cart.sort()               # organise alphabetically

prices = ["45", "60", "40"]   # prices as TEXT (as if from a form)
total = 0
for p in prices:
    total += int(p)           # convert text -> number, then add

print("Cart      :", cart)
print("Items     :", len(cart))
print("Total bill: Rs", total)
```

## App 6 — Marks Editor with a Safe Backup

We add bonus marks to everyone, but keep an untouched **backup** first. This shows
why **cloning** matters: without a copy, the backup would change too.

**Concepts used:** list slicing to clone (`[:]`) · `for` + `range` · indexing · `+=` · mutability

`backup = original[:]` makes an independent copy, so editing `original` is safe:

```python
original = [70, 80, 90]
backup = original[:]          # a real copy (cloning with a slice)

# give everyone 5 bonus marks in the working list
for i in range(len(original)):
    original[i] += 5

print("After bonus:", original)
print("Backup safe:", backup)   # unchanged, thanks to cloning
```

> Reminder: if we had written `backup = original` (no copy), the backup would
> now also show the bonus — because both names would point to the **same** list.

## App 7 — Distance Between Two Points

Stores points as **tuples**, unpacks them, and uses the `math` module to find the
straight-line distance.

**Concepts used:** tuples · tuple unpacking · `import math` · operators & precedence (`**`, `*`, `+`)

Each point is a fixed pair `(x, y)` — a perfect job for a tuple:

```python
import math

point1 = (0, 0)
point2 = (3, 4)

x1, y1 = point1     # unpack the tuple
x2, y2 = point2

distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
print("Distance between points:", distance)
```

## App 8 — Find the Class Topper

Stores each student's marks in a **dictionary**, then loops through to find the
highest.

**Concepts used:** dictionary · `.items()` · `for` loop · `if` + comparison · f-string

Change the marks and see who tops the class:

```python
students = {
    "Rahul": 85,
    "Aisha": 92,
    "John": 78,
}

# show everyone
for name, marks in students.items():
    print(f"{name}: {marks}")

# find the topper
topper = ""
best = 0
for name, marks in students.items():
    if marks > best:
        best = marks
        topper = name

print("-" * 20)
print(f"Topper: {topper} with {best} marks")
```

## App 9 — Word Frequency Counter

Counts how many times each word appears — a real task used in search and text
analysis.

**Concepts used:** string `split` · dictionary · `.get()` with a default · `for` loop · assignment

`counts.get(w, 0)` returns the current count (or 0 if the word is new):

```python
text = "apple banana apple cherry banana apple"

words = text.split()
counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1

print(counts)
```

## App 10 — Student Report Card (Grand Capstone)

One program that uses **strings, a list, a dictionary, loops, conditionals,
operators, and conversion** all together.

**Concepts used:** strings (`upper`, f-string, `*`, field width) · list · dictionary ·
`for` + `range` · `sum` / `len` · `if / elif / else` · conditional expression · comparison

Change `name`, `subjects`, or `marks` to make a new report:

```python
name = "Rahul"
subjects = ["Math", "Science", "English"]
marks = [88, 76, 95]

# build a subject -> marks dictionary
report = {}
for i in range(len(subjects)):
    report[subjects[i]] = marks[i]

total = sum(marks)
average = total / len(marks)

print("REPORT CARD FOR", name.upper())
print("-" * 28)
for subject, m in report.items():
    status = "Pass" if m >= 40 else "Fail"     # conditional expression
    print(f"{subject:10} : {m:3}   ({status})")
print("-" * 28)
print(f"Total   : {total}")
print(f"Average : {average:.1f}")

# decide the overall grade
if average >= 90:
    grade = "A"
elif average >= 75:
    grade = "B"
elif average >= 40:
    grade = "C"
else:
    grade = "Fail"
print(f"Grade   : {grade}")
```
